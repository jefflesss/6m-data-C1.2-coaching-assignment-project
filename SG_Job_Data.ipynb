```python
key_columns = [
    "categories",
    "employmentTypes",
    "metadata_jobPostId",
    "positionLevels",
    "postedCompany_name",
    "salary_type",
    "title"
]
df[key_columns].isna().all(axis=1).sum()
```

```python
df = df.dropna(subset=key_columns, how='all')
df.shape
```

### Exploring the Salary Columns

```python
salary_df = df[
    ["title", "salary_minimum", "salary_maximum", "average_salary"]
].copy()

salary_df = salary_df.sort_values(
    by="salary_maximum",
    ascending=True
)

salary_df.head(20)
```

```python
salary_df.describe()
```

```python
salary_df['salary_range'] = salary_df['salary_maximum'] - salary_df['salary_minimum']

salary_df = salary_df.sort_values(
    by="salary_range",
    ascending=True
)

salary_df.head(20)
```

```python
#looking at the columns in salary minimum and maximum to see if there are any outliers
salary_df[[
    "salary_minimum",
    "salary_maximum",
    "average_salary"
]].describe(percentiles=[0.05, 0.10, 0.25, 0.5, 0.75, 0.90, 0.95, 0.99])
```

```python
#identifying rows where salary_minimum and salary_maximum are equal to 1
salary_df[(salary_df['salary_minimum']==1) & (salary_df['salary_maximum']==1)]
```

```python
#dropped salary_minimum and salary_maximum that are equal to 1
salary_df = salary_df[~((salary_df['salary_minimum']==1) & (salary_df['salary_maximum']==1))]
salary_df.shape
```

```python
salary_df[(salary_df['salary_range']<1000) & (salary_df['salary_range'] != 0)].sort_values(by='salary_minimum', ascending=True).head(20)
```

```python
salary_df[[
    "salary_minimum",
    "salary_maximum",
    "average_salary"
]].describe(percentiles=[0.001, 0.01,0.05, 0.10, 0.25, 0.5, 0.75, 0.90, 0.95, 0.99, 0.999])
```
